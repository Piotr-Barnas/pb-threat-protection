```python
dbutils.widgets.text("catalog", "workspace")
dbutils.widgets.text("schema", "pb_threat_intelligence")
```

```python
catalog = dbutils.widgets.get("catalog").strip()
schema = dbutils.widgets.get("schema").strip()
```

```python
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}""")
```

```python
spark.sql(f"""CREATE VOLUME IF NOT EXISTS {catalog}.{schema}.feed_snapshots""")
```

```python
spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {catalog}.{schema}.feed_observations (
            run_id STRING,
            feed_name STRING,
            category STRING,
            raw_line STRING,
            domain STRING,
            validation_status STRING,
            ingestion_ts TIMESTAMP,
            snapshot_path STRING,
            snapshot_sha256 STRING
        )
    USING DELTA
""")
```

```python

spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {catalog}.{schema}.feed_count_audit (
        run_id STRING,
        run_date DATE,
        feed_name STRING,
        stage STRING,
        status STRING,
        record_count BIGINT,
        previous_count BIGINT,
        ratio DOUBLE,
        logged_at TIMESTAMP
    )
    USING DELTA
""")
```

```python

spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {catalog}.{schema}.domains_classification_current (
        domain STRING,
        category STRING,
        feeds ARRAY<STRING>,
        feed_count INT,
        run_id STRING,
        built_at TIMESTAMP
    )
    USING DELTA
""")
```

```python
spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {catalog}.{schema}.domains_classification_history (
        domain STRING,
        category STRING,
        feeds ARRAY<STRING>,
        feed_count INT,
        valid_from TIMESTAMP,
        valid_to TIMESTAMP,
        is_current BOOLEAN,
        run_id STRING,
        closed_by_run_id STRING
    )
    USING DELTA
""")
```
